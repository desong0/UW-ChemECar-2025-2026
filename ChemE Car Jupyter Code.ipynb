```python
# doc for pyserial: https://www.pyserial.com/docs/reading-data

# Basically, this program will read serial data and plot it in real time,
# Right now it's kinda broken because of the way that bytes are read, but I can fix it pretty easily with some identifier byte but I need to change the arduino code as well

import numpy as np
import serial
import serial.tools.list_ports
import matplotlib.pyplot as plt
from IPython.display import display, clear_output

for port in serial.tools.list_ports.comports():
    print(f"{port.device}: {port.description}")

print("Which COM port will be used? (Windows Only)")
text = input()
print("Using COM"+text)

ser = serial.Serial("COM"+text, 9600, timeout=None)

data = ser.read(1)
print(f"Read {len(data)} bytes: {data}")

fig = plt.figure()
ax = fig.add_subplot(1,1,1)
i = 1
y = np.array([])

# Press the reset button on the arduino to break the loop
# if you need to restart the notebook, you have to close the serial port so you can do that with the cell below this one 
while(True):
    data = ser.read(1)
    print(f"Read {len(data)} bytes: {data}")
    # Read a single byte
    if (data== b'p'):
        break
    x = np.arange(0, i, 1)
    y = np.append(y, int(data))
    ax.set_xlim(0, i)
    ax.cla()
    ax.plot(x, y)
    display(fig)
    clear_output(wait = True)
    plt.pause(0.1)
    i+=1
ser.close()
display(fig)
```
